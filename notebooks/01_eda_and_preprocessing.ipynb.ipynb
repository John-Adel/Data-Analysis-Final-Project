###  To fix this issue 197 trips with no station are dropped. Missing birth year and gender are kept, not filled, and labeled Unknown.

```python
#changing to categories
cat_cols = ['member_gender', 'user_type', 'bike_share_for_all_trip', 'start_station_id', 'start_station_name', 'end_station_id', 'end_station_name']
df[cat_cols] = df[cat_cols].astype('category')
df.info()
```

```python
customer_share = df[df["user_type"] == "Customer"]["member_birth_year"].isnull().sum()
subscriber_share = (
    df[df["user_type"] == "Subscriber"]["member_birth_year"].isnull().sum()
)
shares = pd.DataFrame(
    {
        "no of missing values": [customer_share, subscriber_share],
        "percentage": [
            customer_share / len(df[df["user_type"] == "Customer"]) * 100,
            subscriber_share / len(df[df["user_type"] == "Subscriber"]) * 100,
        ],
    },
    index=["customer_share", "subscriber_share"],
)
shares
```

```python
df['member_gender'] = df['member_gender'].cat.add_categories('Unknown').fillna('Unknown')
```

```python
# dropping the null values in id and names in start and end stations
df = df.dropna(subset=['start_station_id', 'start_station_name', 'end_station_id', 'end_station_name'])
```

```python
customer_share = df[df["user_type"] == "Customer"]["member_birth_year"].isnull().sum()
subscriber_share = (
    df[df["user_type"] == "Subscriber"]["member_birth_year"].isnull().sum()
)
shares = pd.DataFrame(
    {
        "no of missing values": [customer_share, subscriber_share],
        "percentage": [
            customer_share / len(df[df["user_type"] == "Customer"]) * 100,
            subscriber_share / len(df[df["user_type"] == "Subscriber"]) * 100,
        ],
    },
    index=["customer_share", "subscriber_share"],
)
shares
```

```python
## create a function to know the average users ages
def mean_age(age_col_name: str) -> np.float64:
    '''  This function returns the mean of a age column in any data set
    
    
    
    
    Args:
        age_col_name: The name of the age column in the provided dataset 

    Returns:
        The average age 
    
    '''
    try: 
        return df[age_col_name].mean()
    except TypeError:
        print("The age column must be of a numerical Data type")
```
