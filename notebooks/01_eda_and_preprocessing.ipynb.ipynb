###  To fix this issue 197 trips with no station are dropped. Missing birth year and gender are kept, not filled, and labeled Unknown.

```python
#changing to categories
cat_cols = ['member_gender', 'user_type', 'bike_share_for_all_trip']
df[cat_cols] = df[cat_cols].astype('category')
df.info()
```

```python
customer_share = df[df["user_type"] == "Customer"]["member_birth_year"].isnull().sum()
subscriber_share = (
    df[df["user_type"] == "Subscriber"]["member_birth_year"].isnull().sum()
)
shares = pd.DataFrame(
    {
        "no of missing values": [customer_share, subscriber_share],
        "percentage": [
            customer_share / len(df[df["user_type"] == "Customer"]) * 100,
            subscriber_share / len(df[df["user_type"] == "Subscriber"]) * 100,
        ],
    },
    index=["customer_share", "subscriber_share"],
)
shares
```

```python
df['member_gender'] = df['member_gender'].cat.add_categories('Unknown').fillna('Unknown')
```

```python
# dropping the null values in id and names in start and end stations
df = df.dropna(subset=['start_station_id', 'start_station_name', 'end_station_id', 'end_station_name'])
```

```python
df[['start_station_id', 'end_station_id']] = df[['start_station_id', 'end_station_id']].astype('int64')
print(df[['start_station_id', 'end_station_id']].dtypes)
```

# Standardize text columns.

```python
name_cols = ['start_station_name', 'end_station_name']

# Before
for c in name_cols:
    print(c, '- names with extra spaces:', df[c].str.contains(r'\s{2,}').sum())

# Clean: trim the ends and collapse repeated spaces
for c in name_cols:
    df[c] = df[c].str.strip().str.replace(r'\s+', ' ', regex=True)

# After
for c in name_cols:
    print(c, '- names with extra spaces:', df[c].str.contains(r'\s{2,}').sum())
print('unique start names:', df['start_station_name'].nunique())
print('unique end names:', df['end_station_name'].nunique())

# Category check
for c in ['user_type', 'member_gender', 'bike_share_for_all_trip']:
    print(df[c].value_counts(dropna=False), '\n')
```

```python
# Validation:
stations = pd.concat([
    df[['start_station_id', 'start_station_name', 'start_station_latitude', 'start_station_longitude']].set_axis(['id', 'name', 'lat', 'lon'], axis=1),
    df[['end_station_id', 'end_station_name', 'end_station_latitude', 'end_station_longitude']].set_axis(['id', 'name', 'lat', 'lon'], axis=1),
]).drop_duplicates()
print('unique station IDs:', stations['id'].nunique())
print('IDs with more than one name:', (stations.groupby('id')['name'].nunique() > 1).sum())
print('rows in station table:', len(stations))
```

```python
## step 3 ) Good factor for marketing work as in ((`member_birth_year`)) : so need to clean it.
## To know the target audience's age and serve on them
# create a column for users' age depending on the current year
current_year = 2019
df['age'] = current_year - df['member_birth_year']
```

```python
df['age'].dtype

```

```python
## change the new column age data type from float to int
df['age'] = df['age'].astype('Int64')
```

```python
df['age'].dtype
```

```python
## create a function to know the average users ages
def mean_age(age_col_name: str) -> np.float64:
    '''  This function returns the mean of a age column in any data set
    
    
    
    
    Args:
        age_col_name: The name of the age column in the provided dataset 

    Returns:
        The average age 
    
    '''
    try: 
        return df[age_col_name].mean()
    except TypeError:
        print("The age column must be of a numerical Data type")
```

```python
mean_age('age')
```

```python
## another important factor to offer  marketing services is the trip duration time 
## need to change the time from second to minute to be more usable
df['duration_min'] = df['duration_sec'] / 60

```

```python
## I need to change the time from second to hours  to be more usable
df['duration_hour'] = df['duration_sec'] / 3600
```

# Time Features

```python
df['start_time'] = pd.to_datetime(df['start_time'])
df['end_time'] = pd.to_datetime(df['end_time'])
```

```python
print(df[['start_time', 'end_time']].dtypes)
print('start_time range:', df['start_time'].min(), '->', df['start_time'].max())
```

```python
df['start_date'] = df['start_time'].dt.normalize()    
df['start_hour'] = df['start_time'].dt.hour
df['start_day'] = df['start_time'].dt.day               
df['day_of_week'] = df['start_time'].dt.day_name()       
df['day_of_week_num'] = df['start_time'].dt.dayofweek    
df['start_month'] = df['start_time'].dt.month
df['is_weekend'] = df['day_of_week_num'] >= 5
```

```python
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
df['day_of_week'] = pd.Categorical(df['day_of_week'], categories=day_order, ordered=True)
df['day_of_week'].value_counts().sort_index()   # Monday to Sunday order
```

```python
print('end before start:', (df['end_time'] < df['start_time']).sum())
print('start range:', df['start_time'].min(), '->', df['start_time'].max())
print('starts outside Feb:', (df['start_time'].dt.month != 2).sum())
gap = ((df['end_time'] - df['start_time']).dt.total_seconds() - df['duration_sec']).abs().max()
print('max gap vs duration_sec (s):', gap)
print('weekend share:', df['is_weekend'].mean())
```

outliers were found in Q3-Q5 and are removed here.

```python
df['age'] = df['age'].where(df['age'] <= 100, other = np.nan)
df['member_birth_year'] = df['member_birth_year'].where(df['age'] <= 100, other = np.nan)
print(f"ages count: {len(df)}")
print(f"unknown ages: {df[['member_birth_year', 'age']].isna().sum()}")
df = df[df['duration_min'] <= 180].copy()
print(f"duration count: {df['duration_min'].count()}")
```
